##preprocessing

```python
from sklearn.compose import ColumnTransformer,make_column_selector
```

```python
numberic_selector = make_column_selector(dtype_include='number')
categorical_selector = make_column_selector(dtype_exclude='number')
```

```python
numeric_columns=numberic_selector(features)
numeric_columns
```

```python
categorical_columns=categorical_selector(features)
categorical_columns
```

```python
from sklearn.preprocessing import StandardScaler,OrdinalEncoder
```

```python
preprocessing=ColumnTransformer(
    [
        ('numeric',StandardScaler(),numeric_columns),
        ('categorical',OrdinalEncoder(),categorical_columns)
    ]
)
```

```python
preprocessing
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
pipeline = Pipeline([
    ('preprocessing', preprocessing),
    ('model', BaggingClassifier(
        estimator=DecisionTreeClassifier(),
        n_estimators=10,
        random_state=42
    ))
])
pipeline
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'model__n_estimators': [10, 20, 30, 40, 50],
    'model__max_samples': [0.8],
    'model__max_features': [1.0],
    'model__estimator__max_depth': [5, 10],
    'model__estimator__class_weight': ['balanced']
}
```

```python
grid_search = GridSearchCV(pipeline, param_grid, cv=5)
```

```python
grid_search.fit(x_train,y_train)
```

```python
from sklearn.metrics import balanced_accuracy_score,precision_score,recall_score,f1_score
```

```python
best_model=grid_search.best_estimator_
y_train_pred_grid= best_model.predict(x_train)
```

```python
balanced_accuracy=balanced_accuracy_score(y_train,y_train_pred_grid)
precision=precision_score(y_train,y_train_pred_grid,pos_label='yes')
recall=recall_score(y_train,y_train_pred_grid,pos_label='yes')
f1_score_value=f1_score(y_train,y_train_pred_grid,pos_label='yes')

balanced_accuracy
```
